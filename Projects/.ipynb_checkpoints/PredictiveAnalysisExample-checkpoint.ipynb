```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/ageron/handson-ml/master/datasets/housing/housing.csv")
df
```

```python
df.dropna(inplace=True)
```

```python
sns.pairplot(df[["median_house_value", "median_income", "total_rooms", "housing_median_age", "population"]])
plt.show()
```

```python
# Correlation matrix to see the relationships between variables
correlation_matrix = df[["median_house_value", "median_income", "total_rooms", "housing_median_age"]].corr()
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm")
```

### Observations from EDA
median_house_value(target variable) has a string positive correlation with median_income, suggesting higher incomes lead to higher home prices.
We'll use the median_income, total_rooms, and housing_median_age as predictors for housing prices.

### Steps for building a Predictive Analysis Model

1) Prepare your data
2) Train your model
3) Analyze the Performance of your model
4) Test you model on the new data.

```python
# Prepare your data
x = df[["median_income", 'total_rooms', 'housing_median_age']]
x
```

```python
y = df['median_house_value']
y
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.2, random_state=42)
```

```python
# Train you model
model = LinearRegression()
```

```python
model
```

```python
# Rrain your model
model.fit(x_train, y_train)
```

```python
#Test the models performance
y_pred = model.predict(x_test)
```

```python
pd.Series(y_pred)
```

```python
y_test
```

```python
# Measure the Accuracy of Our Model
mse = mean_squared_error(y_test,y_pred)
r2 = r2_score(y_test,y_pred)
print(f"Mean Squared Error: {mse}")
print(f"R-squared: {r2}")
```

```python
# Test the model on a Brand New House
example_house = pd. DataFrame({
"median_income": [4.5], # Example median income
"total_rooms": [2000], # Example total rooms
"housing_median_age": [15] # Example age of the house in years
})
```

```python
model.predict(example_house)
```
