### plot customization

```python
years = [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021]

income = [55, 56, 62, 61, 72, 72, 73, 75]
```

```python
income_ticks = list(range(50, 81, 2))
```

```python
plt.plot(years, income)
plt.title('Income of Charles')
plt.xlabel('Years')
plt.ylabel('income in thousands')
plt.yticks(income_ticks, [f"${x}k USD" for x in income_ticks])
```

### Legend and multiple plots

```python
stock_a = [100, 102, 99, 101, 101, 100, 102]
stock_b = [90, 95, 102, 104, 105, 103, 109]
stock_c = [110, 115, 100, 105, 100, 98, 95]
```

```python
plt.plot(stock_a, label = 'Google')
plt.plot(stock_b, label = 'Amazon')
plt.plot(stock_c, label = 'Apple')
plt.legend(loc= 'lower right')
```

```python
years = [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021]

income = [55, 56, 62, 61, 72, 72, 73, 75]
```

```python
plt.pie(income, autopct="%.2f%%", labels = None)
plt.legend(years,  loc= 'lower right')
```

### plot styling (from matplotlib import style)

```python
years = [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021]

income = [55, 56, 62, 61, 72, 72, 73, 75]
plt.pie(income, autopct="%.2f%%", labels = None)
plt.legend(years,  loc= 'lower right')
style.use('ggplot')
style.use('dark_background')
```

### subplots

```python
 x=  np.arange(100)
```

```python
fig, axis= plt.subplots(2,2)
axis[0,0].plot(x, np.sin(x))
axis[0,0].set_title('sine wave')

axis[0,1].plot(x, np.sin(x))
axis[0,1].set_title('Cosine wave')

axis[1,0].plot(x, np.sin(x))
axis[1,0].set_title('Random Function')

axis[1,1].plot(x, np.log(x))
axis[1,1].set_title('Log Function')
axis[1,1].set_xlabel('Test')

fig.suptitle("Four Plots")

plt.tight_layout()
```

```python
plt.savefig("fourplot.png", dpi= 300, transparent=False, bbox_inches='tight')
```

### 3D Plotting

```python
ax = plt.axes(projection="3d")

x = np.random.random(100)
y = np.random.random(100)
z = np.random.random(100)

ax.scatter(x, y, z)
ax.set_title("3d plot ")
ax.set_xlabel("Test")
```

```python
ax = plt.axes(projection="3d")

x = np.arange(0, 50, 0.1)
y = np.sin(x)
z = np.cos(x)

ax.plot(x, y, z)
ax.set_title("3d plot ")
ax.set_xlabel("Test")
```

```python
ax = plt.axes(projection="3d")

x = np.arange(0, 50, 0.1)
y = np.arange(0, 50, 0.1)
z = np.cos(x + y)

ax.plot(x, y, z)
ax.set_title("3d plot ")
ax.set_xlabel("Test")
```

```python
ax = plt.axes(projection="3d")

x = np.arange(-5, 5, 0.1)
y = np.arange(-5, 5, 0.1)

X, Y = np.meshgrid(x,y)

Z = np.sin(X) * np.cos(Y)



ax.plot_surface(X, Y, Z, cmap= "Spectral")
ax.set_title("3d plot ")
ax.set_xlabel("Test")
ax.view_init(azim=0, elev=90)
```
